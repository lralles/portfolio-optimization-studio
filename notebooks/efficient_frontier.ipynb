```python
import sys
sys.path.insert(0, '..')

from src.analysis.efficient_frontier.analysis import compute_efficient_frontier
from src.analysis.efficient_frontier.plot import plot_efficient_frontier
from src.consumers.consume_bovespa import consume_bovespa
from src.consumers.consume_ifix import consume_ifix
from src.consumers.consume_sp500brl import consume_sp500brl
from src.consumers.consume_imab import consume_imab
from src.consumers.consume_irfm import consume_irfm
from src.consumers.consume_imas import consume_imas
import os
os.chdir('..')
```

```python
bovespa = consume_bovespa()
ifix = consume_ifix()
sp500brl = consume_sp500brl()
imab = consume_imab()
irfm = consume_irfm()
imas = consume_imas()

indexes = [bovespa, ifix, sp500brl, imab, irfm, imas]
risk_free_rate = 0.1075

results = compute_efficient_frontier(indexes, risk_free_rate=risk_free_rate)
plot_efficient_frontier(results, 'Efficient Frontier')
```

```python
import pandas as pd

print("=== Min Variance Portfolio ===")
print(f"Return: {results['min_variance_portfolio']['annualized_return']:.2f}%")
print(f"Volatility: {results['min_variance_portfolio']['volatility']:.2f}%")
print("Weights:")
print(pd.Series(results['min_variance_portfolio']['weights']).to_string())

print("\n=== Max Return Portfolio ===")
print(f"Return: {results['max_return_portfolio']['annualized_return']:.2f}%")
print(f"Volatility: {results['max_return_portfolio']['volatility']:.2f}%")
print("Weights:")
print(pd.Series(results['max_return_portfolio']['weights']).to_string())

print(f"\n=== Tangency Portfolio (rf={results['risk_free_rate']:.2f}%) ===")
print(f"Return: {results['tangency_portfolio']['annualized_return']:.2f}%")
print(f"Volatility: {results['tangency_portfolio']['volatility']:.2f}%")
print("Weights:")
print(pd.Series(results['tangency_portfolio']['weights']).to_string())
```
