```python
import sys
sys.path.insert(0, '../..')

import os
os.chdir('../..')

from src.consumers.consume import consume
from src.portfolio_generator.inflation_bonds_indexes import INFLATION_BONDS_INDEXES, generate_inflation_bonds_portfolios
from src.portfolio_generator.inflation_bonds_indexes import simulate_inflation_bonds_portfolio_monthly, simulate_inflation_bonds_portfolio
from src.analysis.risk_return_analysis.analysis import compute_risk_return_analysis
from src.analysis.risk_return_analysis.plot import plot_portfolio_risk_return
from src.analysis.risk_return_analysis.type import print_results
```

```python
indexes = {name: consume(name) for name in INFLATION_BONDS_INDEXES}
portfolios = generate_inflation_bonds_portfolios(indexes, random_portfolios=40, seed=42)
print(f'Generated {len(portfolios)} portfolios from {len(INFLATION_BONDS_INDEXES)} indexes.')
```

```python
simulated_portfolios = []
for name, composition in portfolios:
    portfolio = simulate_inflation_bonds_portfolio_monthly(composition, name)
    simulated_portfolios.append(portfolio)

baseline_indexes = list(indexes.values())
all_results = baseline_indexes + simulated_portfolios
results = compute_risk_return_analysis(all_results)
plot_portfolio_risk_return(results, 'Risk-Return: Inflation Bond Portfolios and Baselines')
print_results(results)

buy_hold_portfolios = []
for name, composition in portfolios:
    portfolio = simulate_inflation_bonds_portfolio(composition, f'Buy & Hold {name}')
    buy_hold_portfolios.append(portfolio)

buy_hold_results = compute_risk_return_analysis(baseline_indexes + buy_hold_portfolios)
plot_portfolio_risk_return(buy_hold_results, 'Risk-Return: Buy-and-Hold Inflation Bond Portfolios and Baselines')
print_results(buy_hold_results)
```
