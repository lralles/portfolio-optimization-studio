```python
bovespa = consume('bovespa')
sp500brl = consume('sp500brl')
imab = consume('imab')
imas = consume('imas')

indexes = [bovespa, imab, imas, sp500brl]

test_portfolio = {
    'bovespa': 20,
    'imab': 40,
    'imas': 10,
    'sp500brl': 30
}

results = compute_efficient_return_region(
    indexes,
    test_portfolio=test_portfolio,
    window_size=10 * 252,
    sample_number=20,
    independent_windows=False,
    n_points=200
)

plot_efficient_return_region(results, 'Efficient Return Region Analysis - Independent Windows')
print_results(results)
```

```python
results_overlapping = compute_efficient_return_region(
    indexes,
    test_portfolio=test_portfolio,
    window_size=252,
    sample_number=10,
    independent_windows=False,
    n_points=200
)

plot_efficient_return_region(results_overlapping, 'Efficient Return Region Analysis - Overlapping Windows')
print_results(results_overlapping)
```
